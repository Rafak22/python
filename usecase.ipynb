```python
%%writefile app.py

import os
import chainlit as cl

from docling.document_converter import DocumentConverter
from langchain_docling import DoclingLoader
from langchain_docling.loader import ExportType
from langchain_community.vectorstores import Chroma
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores.utils import filter_complex_metadata
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate

OPENROUTER_API_KEY = os.environ.get("OPENROUTER_API_KEY", "")
EMBED_MODEL_ID = "sentence-transformers/all-MiniLM-L6-v2"


def build_rag_chain(vector_store):
    llm = ChatOpenAI(
        model="nvidia/nemotron-3-nano-30b-a3b:free",
        temperature=0,
        base_url="https://openrouter.ai/api/v1",
        api_key=OPENROUTER_API_KEY,
    )

    retriever = vector_store.as_retriever(search_kwargs={"k": 4})

    prompt_template = """You are a helpful study assistant. Use the following
excerpts from a university lecture or textbook to answer the student's question.
If the answer isn't in the excerpts, say "I couldn't find that in the document."

Context:
{context}

Student's Question: {question}

Answer:"""

    PROMPT = PromptTemplate(
        template=prompt_template,
        input_variables=["context", "question"]
    )

    def run_chain(query):
        docs = retriever.invoke(query)
        context = "\n\n".join([doc.page_content for doc in docs])
        final_prompt = PROMPT.format(context=context, question=query)
        response = llm.invoke(final_prompt)
        return {"result": response.content, "source_documents": docs}

    return run_chain


@cl.on_chat_start
async def on_chat_start():
    files = await cl.AskFileMessage(
        content="👋 Welcome! Please upload your lecture PDF to get started.",
        accept=["application/pdf"],
        max_size_mb=20,
    ).send()

    file = files[0]

    msg = cl.Message(content="📄 Parsing PDF with Docling... please wait.")
    await msg.send()

    tmp_path = file.path

    loader = DoclingLoader(
        file_path=tmp_path,
        export_type=ExportType.DOC_CHUNKS,
    )
    chunks = loader.load()

    msg.content = "🔍 Building search index..."
    await msg.update()

    filtered_chunks = filter_complex_metadata(chunks)
    embeddings = HuggingFaceEmbeddings(model_name=EMBED_MODEL_ID)
    vector_store = Chroma.from_documents(
        documents=filtered_chunks,
        embedding=embeddings,
        collection_name="study_assistant"
    )

    chain = build_rag_chain(vector_store)
    cl.user_session.set("chain", chain)

    msg.content = f"✅ **{file.name}** loaded! Ask me anything about it."
    await msg.update()


@cl.on_message
async def on_message(message: cl.Message):
    chain = cl.user_session.get("chain")

    if chain is None:
        await cl.Message(content="⚠️ Please upload a PDF first!").send()
        return

    msg = cl.Message(content="🤔 Thinking...")
    await msg.send()

    result = chain(message.content)
    answer = result["result"]
    sources = result["source_documents"]

    source_text = "\n\n".join([
        f"**Excerpt {i+1}:** {doc.page_content}"
        for i, doc in enumerate(sources)
    ])

    msg.content = f"{answer}\n\n---\n📌 **Sources:**\n{source_text}"
    await msg.update()
```
