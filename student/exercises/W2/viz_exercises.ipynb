### 1. Antibiotic use in children

The bar plot and the pie chart below show the distribution of pre-existing medical conditions of children involved in a study on the optimal duration of antibiotic use in treatment of tracheitis, which is an upper respiratory infection.

![](../../assets/viz_exercise_1.png)

a. What features are apparent in the bar plot but not in the pie chart?

b. What features are apparent in the pie chart but not in the bar plot?

c. Which graph would you prefer to use for displaying these categorical data?


SOL


### a. Features apparent in the bar plot but NOT in the pie chart

The bar plot clearly shows the **exact counts** for each condition, making it easy to compare the differences between categories. It is also much easier to see the **ranking** of conditions from most to least common, and the **differences between similar-sized categories** (like Gastrointestinal vs Immunocompromised) are much clearer.

---

### b. Features apparent in the pie chart but NOT in the bar plot

The pie chart shows the **proportion/percentage** of each condition relative to the whole, making it easier to see that Prematurity takes up roughly a third of all cases at a glance.

---

### c. Which graph is preferred for these data?

The **bar plot** is preferred for these categorical data because there are 8 categories, and comparing their sizes is much easier with bar lengths than with pie slices. Pie charts become hard to read when there are more than 4-5 categories, especially when some slices are very similar in size (like Gastrointestinal and Immunocompromised).

### 2. Black Lives Matter.

A Washington Post-Schar School poll conducted in the United States in June 2020, among a random national sample of 1,006 adults, asked respondents whether they support or oppose protests following George Floyd’s killing that have taken place in cities across the US. The survey also collected information on the age of the respondents. (Washington Post 2020) The results are summarized in the stacked bar plot below.

![](../../assets/viz_exercise_3.png)

a. Based on the stacked bar plot, do views on the protests and age appear to be associated? Explain your reasoning.

b. Conjecture other possible variables that might explain the potential association between these two variables.

### a. Do views on the protests and age appear to be associated?

Yes, there is a clear association between age and views on the protests. Younger age groups (18-29 and 30-39) show a much larger proportion of "Strongly support" (dark teal) compared to older age groups. In contrast, older respondents (65+) show a noticeably larger proportion of "Strongly oppose" (dark orange). This pattern suggests that as age increases, support for the protests tends to decrease and opposition tends to increase.

---

### b. Other possible variables that might explain this association

There are several other variables that could explain the relationship between age and opinion:

- **Political affiliation** — younger people tend to lean more liberal, which may explain higher support regardless of age itself
- **Education level** — younger generations on average have higher education levels, which may influence opinions on social justice issues
- **Media consumption** — younger people consume more social media, which may expose them to different perspectives than older generations who rely more on traditional media
- **Personal experience** — younger people are more likely to have personally experienced or witnessed racial discrimination, which may shape their views

### 3. Heart transplant data display.

The Stanford University Heart Transplant Study was conducted to determine whether an experimental heart transplant program increased lifespan. Each patient entering the program was officially designated a heart transplant candidate, meaning that they were gravely ill and might benefit from a new heart. Patients were randomly assigned into treatment and control groups. Patients in the treatment group received a transplant, and those in the control group did not. The visualizations below display two different versions of the study results.9 (Turnbull, Brown, and Hu 1974)

![](../../assets/viz_exercise_5.png)

a. Provide one aspect of the two group comparison that is easier to see from the stacked bar plot (left)?

b. Provide one aspect of the two group comparison that is easeir to see from the standardized bar plot (right)?

c. For the Heart Transplant Study which of those aspects would be more important to display? That is, which bar plot would be better as a data visualization?

### a. Easier to see from the stacked bar plot (left — Count)

The **actual number of patients** in each group is easier to see from the left plot. You can clearly tell that the treatment group had many more patients (around 69) compared to the control group (around 34), which shows the **difference in group sizes**.

---

### b. Easier to see from the standardized bar plot (right — Proportion)

The **survival rate comparison** between the two groups is easier to see from the right plot. It is much clearer that the treatment group had a higher proportion of alive patients (blue) compared to the control group, because both bars are scaled to 100% making a fair comparison possible.

---

### c. Which plot is better for this study?

The **standardized bar plot (right)** is better for this study because the main research question is whether the transplant **increased survival**, not how many patients were in each group. Since the two groups have different sizes, comparing raw counts would be misleading. The proportion plot gives a fair comparison by showing the survival rate of each group regardless of group size.

### 4. Meat consumption and life expectancy.

In data collected for You et al. (2022), total meat intake is associated with life expectancy (at birth) in 175 countries. Meat intake is measured in kg per capita per year (averaged over 2011 to 2013). The two ridge plots show an association between income and meat consumption (higher income countries tend to eat more meat) and an association between income and life expectancy (higher income countries have higher life expectancy).

![](../../assets/viz_exercise_7.png)

a. Do the graphs above demonstrate that meat consumption and life expectancy are associated? That is, can you tell if countries with low meat consumption have low life expectancy? Explain.

b. Let’s assume that you had a plot comparing meat consumption and life expectancy, and they do seem associated. Your friend says that the plot shows that high meat consumption leads to a longer life. You correctly say, no, we can’t tell if there is a causal realtionship because the relationship is confounded by income level. Explain what you mean.

c. How can you investigate the relationship between meat consumption and life expectancy in the presence of confounding variables (like income)?

### a. 

The two graphs **cannot directly** show the association between meat consumption and life expectancy because each graph only shows one variable against income. However, taken together they suggest an indirect link — low income countries have both low meat consumption AND low life expectancy, while high income countries have both high meat consumption AND high life expectancy. So the two variables appear associated, but only **through income**, not directly.

---

### b. 
When we say the relationship is **confounded by income**, we mean that income level affects both variables at the same time. Richer countries can afford more meat AND also have better healthcare, cleaner water, and better living conditions — all of which increase life expectancy. So it is not the meat itself causing longer life, but rather **income is the real driver** of both. We cannot separate the effect of meat from the effect of income just by looking at these plots.

---

### c. 

There are two main approaches:

- **Statistical control** — use multiple regression to include income as a variable, so we can see the effect of meat consumption on life expectancy while holding income constant
- **Visual control** — create a scatter plot of meat consumption vs life expectancy but **color the points by income group**, so we can see if the association still exists within each income group separately
