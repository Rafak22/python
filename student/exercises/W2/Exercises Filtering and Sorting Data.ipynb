### Step 4. How many products cost more than $10.00?

```python
chipo.head(20)
```

```python
chipo.sort_values(by=['item_name'], ascending=False)
#you can see here that the item price is based on both the qauntity and the choice_description
```

```python
chipo["item_price"] = chipo["item_price"].str.replace("$", "", regex=False).astype(float)
```

```python
chipo["product_price"] = chipo["item_price"] / chipo["quantity"]
chipo
```

```python
chipo.loc[(chipo["choice_description"] == '[Diet Coke]') & (chipo["item_name"] == "Canned Soda")]
filtered_chipo=chipo.drop_duplicates(['item_name','choice_description'])
filtered_chipo = filtered_chipo.loc[ filtered_chipo["product_price"]>10.0 , ["item_name","choice_description","product_price"] ].reset_index(drop=True)

print(filtered_chipo.shape[0])
```

```python
# ----------- THIS was MY ANSWER (WRONG) -----------
# The mistake here is that I only filtered by item_price > 10,
# which represents the price of a SINGLE item, not the total order price.
#
# I also used nunique() on item_name, which counts only unique item names
# and ignores choice_description (different customizations of the same item).
#
# However, in this dataset:
# - The actual price should be calculated using: quantity * item_price
# - Each combination of item_name + choice_description represents a different product
#
# So this approach:
# 1. Ignores quantity (total_price is wrong)
# 2. Ignores choice_description (different items treated as the same)
# 3. Counts unique names instead of actual products
#
# That's why the result (31) is incorrect.
# -----------------------------------------------
```
