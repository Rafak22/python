```python
#task2

from docling.document_converter import DocumentConverter, PdfFormatOption
from docling.datamodel.pipeline_options import PdfPipelineOptions
from docling.datamodel.base_models import InputFormat

pipeline_options = PdfPipelineOptions()
pipeline_options.do_ocr = False
pipeline_options.do_table_structure = True

converter = DocumentConverter(
    format_options={
        InputFormat.PDF: PdfFormatOption(pipeline_options=pipeline_options)
    }
)

result = converter.convert("https://arxiv.org/pdf/2408.09869")
print(result.document.export_to_markdown())
```

```python

#task4

from docling.document_converter import DocumentConverter, WordFormatOption
from docling.datamodel.base_models import InputFormat

converter = DocumentConverter(
    allowed_formats=[
        InputFormat.PDF,
        InputFormat.DOCX,
        InputFormat.HTML,
        InputFormat.PPTX,
    ]
)

sources = [
    "https://arxiv.org/pdf/2408.09869",          # PDF
    "https://www.w3.org/WAI/WCAG21/wcag.html",   # HTML
]

results = converter.convert_all(sources)

for res in results:
    print("=== FORMAT:", res.input.format)
    print(res.document.export_to_markdown()[:500])
    print()
```

```python
converter = DocumentConverter()
result = converter.convert("https://arxiv.org/pdf/2408.09869")

confidence = result.confidence

# Calculate overall confidence from page-level confidences
if confidence.pages:
    # Filter out elements that are not objects or don't have a 'score' attribute
    valid_page_confs = [p for p in confidence.pages if hasattr(p, 'score')]

    if valid_page_confs:
        total_score = sum(p.score for p in valid_page_confs)
        overall_score = total_score / len(valid_page_confs)

        # Simple heuristic for grade based on average score
        if overall_score > 0.9:
            overall_grade = "Excellent"
        elif overall_score > 0.7:
            overall_grade = "Good"
        elif overall_score > 0.5:
            overall_grade = "Moderate"
        else:
            overall_grade = "Low"

        print("Overall confidence grade :", overall_grade)
        print("Overall confidence score :", f"{overall_score:.3f}")
    else:
        print("Overall confidence grade : N/A (no valid pages with scores found)")
        print("Overall confidence score : N/A (no valid pages with scores found)")
else:
    print("Overall confidence grade : N/A (no pages found)")
    print("Overall confidence score : N/A (no pages found)")
print()

# Per-page confidence (apply same filtering for consistency and robustness)
for i, page_conf in enumerate(confidence.pages):
    if hasattr(page_conf, 'grade') and hasattr(page_conf, 'score'):
        print(f"Page {i+1}: grade={page_conf.grade}  score={page_conf.score:.3f}")
    else:
        print(f"Page {i+1}: Invalid page confidence object or missing attributes: {page_conf}")
```
